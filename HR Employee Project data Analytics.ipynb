## Task 1 - Build the three DataFrames and make them as CSV files

```python
# Importing libraries

import pandas as pd
import numpy as np

# Creating Project Dataframe

project_df = pd.DataFrame({ 
    "ID" : ['A001', 'A002', 'A003', 'A004', 'A005', 'A002', 'A005', 'A003',
            'A001', 'A003', 'A001', 'A004', 'A004', 'A005'], # employee ID for each project
    "Project": ["Project 1", "Project 2", "Project 3", "Project 4", "Project 5", "Project 6",
                "Project 7", "Project 8", "Project 9", "Project 10", "Project 11", "Project 12", "Project 13", "Project 14"], # project name
    "Cost" : [1002000, 2000000, 4500000, 5500000, np.nan, 680000, 400000, 
              350000, np.nan, 300000, 2000000, 1000000, 3000000, 200000], # project cost, 2 missing values
    "Status" : ["Finished", "Ongoing", "Finished", "Ongoing", "Finished", "Failed", "Finished", 
                "Failed", "Ongoing", "Finished", "Failed", "Ongoing", "Finished", "Finished"] # project status
})

# Creating Employee Dataframe

employee_df = pd.DataFrame({
    "ID" : ['A001', 'A002', 'A003', 'A004', 'A005'],                                        # employee ID column
    "Name" : ["John Alter", "Alice Luxumberg", "Tom Sabestine", "Nina Adgra", "Amy Johny"], #employee full name column
    "Gender" : ["M", "F", "M", "F", "F"],                                                   #gender column
    "City" : ["Paris", "London", "Berlin", "Newyork", "Madrid"],                            #city column
    "Age" : [25, 27, 29, 31, 30]                                                            # age column
})       

# Creating Seniority Level Dataframe

seniority_df = pd.DataFrame({
    "ID" : ['A001', 'A002', 'A003', 'A004', 'A005'], # employee ID column, used for joining
    "Designation Level": [2, 2, 3, 2, 3]             # designation level column                           
})                            
                            
# save the Project dataframe to csv file
project_df.to_csv("Project.csv", index=False)

# save the Employee dataframe to csv file
employee_df.to_csv("Employee.csv", index=False)

# save the Seniority dataframe to csv file
seniority_df.to_csv("Seniority.csv", index=False)


# print the Project dataframe, including the missing costs values
print(project_df)

# print the Employee dataframe 
print(employee_df)

# print the Seniority
print(seniority_df)
```

## Task 2 - Filling the missing cost values with running average (using For loop)

```python
# Read the Project DataFrame from the CSV saved in Task 1
project_df = pd.read_csv("Project.csv")

# Converting the 'Cost' column to numeric so we can calculate average
project_df["Cost"] = pd.to_numeric(project_df["Cost"])

# Loop through the rows to fill missing Cost values using running average
for i in range(len(project_df)):
    # If the current Cost value is NaN ( missing)
    if pd.isna(project_df.loc[i, "Cost"]):
        # Calculate running average from all previous non-missing Cost values
        running_avg = project_df.loc[:i-1, "Cost"].dropna().mean()
        # Fill the missing value with the computed running average
        project_df.loc[i, "Cost"] = running_avg

# save the updated project dataframe (with missing values filled) back to the csv file
project_df.to_csv("Project.csv", index=False)

# Display the updated DataFrame with filled Cost values
project_df

```

## Task 3 - Split Name Column in Employee DataFrame into First Name and Last Name

```python
# Read the Project DataFrame from the CSV saved in Task 1
employee_df = pd.read_csv("Employee.csv")

# Split the 'Name' column into two new columns 'FirstName' and 'LastName'
employee_df[["First Name", "Last Name"]] = employee_df["Name"].str.split(' ', n=1, expand=True)

# Drop the original 'Name' column as it is replaced
employee_df = employee_df.drop(columns = ["Name"])

# save the updated employee dataframe back to the csv file
employee_df.to_csv("Employee.csv", index=False)


# Display the updated DataFrame
print(employee_df)
```

## Task 4 - Joining all the DataFrames in to one - Final

```python
# Read all three updated CSV files

project_df = pd.read_csv("Project.csv")
employee_df = pd.read_csv("Employee.csv")
seniority_df = pd.read_csv("Seniority.csv")

# Merge project_df with employee_df on 'ID'
merged_df = pd.merge(project_df, employee_df, on='ID')

# Merge the result with seniority_df on 'ID'
Final = pd.merge(merged_df, seniority_df, on='ID')

# Display the combined DataFrame
Final

```

## Task 5 - Adding a Bonus Column to Final DataBase - 5% of Cost for finished Products

```python
# Define a new function to calculate Bonus

def Calculate_bonus(row):
    if row['Status'] == 'Finished': # Check if status is finished 
        return row['Cost'] * 0.05   # Calculate 5% of project cost
    else:
         return 0 # Id Status is unfinished
        
# Apply function to each row of Final DataBase
Final['Bonus'] = Final.apply(Calculate_bonus, axis=1)

# Display updated Final DataBase
Final
```

## Task 6 - Demote Designation Level by 1 for failed projects; delete anyone above level 4


```python
# Demote by 1 level where project status is 'Failed'
Final.loc[Final['Status'] == 'Failed', 'Designation Level'] += 1

# Drop employees with Designation Level > 4
Final = Final[Final['Designation Level'] <= 4]

# Display Final DataFrame with required columns to confirm
Final 
```

## Task 7 - Add 'Mr' or 'Mrs' to  first name and drop Gender Column

```python
# Adding Prefix to First Name

for i in Final.index:
     if Final.loc[i, "Gender"] == "M":  
         Final.loc[i, "First Name"] = "Mr. " + Final.loc[i, "First Name"] # add the "Mr." title in front of the first name
     else:
         Final.loc[i, "First Name"] = "Mrs. " + Final.loc[i, "First Name"] # add "Mrs." in front of the first name for non-male records

# drop the gender column 
Final = Final.drop(columns=["Gender"])

# Display updated Final DataFrame
Final
```

## Task 8 - Promote Designation Level by 1 for employees older than 29 using IF

```python
# Loop through every row for Final DataFrame using 

for i in Final.index:
    if Final.loc[i, 'Age'] > 29: # Step 2: Check if Age > 29
        Final.loc[i, 'Designation Level'] -= 1  # Increase Designation Level by 1

# Display updated Final DataFrame
Final
```

## Task 9 - Create \"TotalProjCost\" for each Employee

```python
# group the Final dataframe by ID and First Name, then add up the Cost column

TotalProjCost = Final.groupby(["ID", "First Name"])["Cost"].sum().reset_index()

# rename the summed "Cost" column to "Total Cost" for clarity

TotalProjCost = TotalProjCost.rename(columns={"Cost": "Total Cost"})

# Display the new TotalProjCost dataframe with required columns
TotalProjCost
```

## Task 10 - Print Employees whose City name contains letter \"o\"

```python
# Filter rows where 'City' contains the letter 'o'
employees_with_o_city = Final[Final['City'].str.contains('o', case=False, na=False)]

# Print the updated result
employees_with_o_city
```
